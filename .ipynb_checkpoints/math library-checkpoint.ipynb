```python
import math
#constants
print(math.pi)
print(math.tau)
print(math.e)
print(math.inf)
print(math.nan)
```

```python
#square root, power, abosolute of number
a = 4
print(math.sqrt(a))
print(math.pow(a,3))
print(abs(a))
```

```python
#round up and down of two numbers or ciel and florr
a=5.3
b=6.3
print(math.ceil(a))
print(math.floor(b))

if a==b:
     print("both are equal also")
```

```python
#taking lcm and hcf
c=9
d=6
print(math.gcd(c,d))
print(math.lcm(c,d))
```

```python
#trignometric functions using math library 
θ = math.pi/2
print(round(math.sin(θ), 10))
print(round(math.cos(θ), 10))
print(round(math.tan(θ), 10))
#the output for tan is a lil bit confusing
```

```python
#degrees to radians and radians to degrees
print(math.radians(180))
print(math.degrees(math.pi))
```

```python
#logarithm and exponential
print(math.log(10)) #natural log
print(math.log10(10)) #base 10
print(math.exp(1))
```

```python
#find remainder or modulus
print(math.fmod(20,4))
```

```python
#distance between two points                    
print(math.dist((2,5),(3,4)))
```

```python
#finding hypotenuse using pythagorus
print(math.hypot(3,4))
```

```python
print(math.isclose(2.7+0.3,3))
```

```python
print(math.trunc(8.909))
```

```python
print(math.isinf(math.inf))
```
