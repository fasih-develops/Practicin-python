```python
import tkinter as tk

def add_numbers():
    num1 = float(entry1.get())
    num2 = float(entry2.get())
    result_label.config(text=f"Sum: {int(num1 + num2)}")

def clear_fields():
    entry1.delete(0, tk.END)
    entry2.delete(0, tk.END)
    result_label.config(text="Result will be displayed here")

root = tk.Tk()
root.title("Shabash")

entry1 = tk.Entry(root, width=10)
entry1.grid(row=0, column=0, padx=5, pady=5)

entry2 = tk.Entry(root, width=10)
entry2.grid(row=0, column=1, padx=5, pady=5)

add_button = tk.Button(root, text="Add", command=add_numbers)
add_button.grid(row=1, column=1, padx=5, pady=5)

clear_button = tk.Button(root, text="Clear", command=clear_fields)
clear_button.grid(row=1, column=2, padx=5, pady=5)

result_label = tk.Label(root, text="Result will be displayed here")
result_label.grid(row=2, column=1, columnspan=2, padx=5, pady=5)

root.mainloop()
```
