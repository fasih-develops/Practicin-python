```python
import statistics as st
```

### Measures of central tendancy

```python
x= [2,3,4,5,6]
y= [4,6,8,7,9]
#to find mean of a given set of data
l = st.mean(x)
print(l)
#to find precise mean of a given data
print(st.fmean(x))
print(st.geometric_mean(x))
print(st.harmonic_mean(x))
print(st.mode(x))
print(st.median(x))
print(st.median_low(x))
print(st.median_high(x))
print(st.multimode(x))
print(st.median_grouped(x))
```

### Measures of spread

```python
#variance is the measure of average squared distance of each data point from the mean, it tells how much the values are spread out
print(st.variance(x))
#standard deviation is the square root of the variance
print(st.stdev(x))
#population variance describes how much each value in a population differs from the mean population
print(st.pvariance(x))
#standard deviation of the entire population
print(st.pstdev(x))
```

### Relationship between two data sets

```python
#Correlation is used to measure the strength and direction of the relationship bw two variables 
print(st.correlation(x,y))
#covariance tells us whether two variables tend to increase or decrease together
print(st.covariance(x,y))
#linear regression is used to find the best straight line relationship between two variables and use that relationship to predict one variable from another
print(st.linear_regression(x,y))
```

### Normal distribution

```python
#normal distributon is a class of st. in python used to work with the normal(gaussian) probability distribution
dist = st.NormalDist(10,37)
#it is used to find mean given above
print(dist.mean)
#it is used to find standard devaition given above
print(dist.stdev)
#Probability Density Function, it gives the probability density at a specific value
print(dist.pdf(30))
#Cumulative Distribution Functuion, which is a randomly selected value which is less than or equal to x
print(dist.cdf(30))
#opposite of cdf, it takes the probability and find the value corresponding to that probability, it must be bw 0.1 - 1
print(dist.inv_cdf(0.2))
#generates n random samples
print(dist.samples(5))
#a z-score tells how many standard devaitions a value is away from the mean value
print(dist.zscore(10))
```

### Quantiles


```python
#it is used to give data into equal sized intervals, n= number of intervals here 2 to 3, 3 to 5 and 5 to 6
print(st.quantiles(x,n=3))
```
