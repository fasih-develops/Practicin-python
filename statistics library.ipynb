```python
import statistics as st
```

## Measures of central tendancy

```python
x= [2,3,4,5,6]
#to find mean of a given set of data
l = st.mean(x)
print(l)
#to find precise mean of a given data
print(st.fmean(x))
print(st.geometric_mean(x))
print(st.harmonic_mean(x))
print(st.mode(x))
print(st.median(x))
print(st.median_low(x))
print(st.median_high(x))
print(st.multimode(x))
print(st.median_grouped(x))
```

## Measures of spread

```python
#variances
#variance is the measure of average squared distance of each data point from the mean, it tells how much the values are spread out
print(st.variance(x))
#standard deviation is the square root of the variance
print(st.stdev(x))
#population variance describes how much each value in a population differs from the mean population
print(st.pvariance(x))
#standard deviation of the entire population
print(st.pstdev(x))
```
