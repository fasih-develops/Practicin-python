```python
import random
#returns a random number betweeen 0.1 and 1 by default
print(random.random())
```

```python
#returns a random  decimal number betweeen a given interval (a,b)
print(random.uniform(2,10))
```

```python
#returns a random integer between a given interval (a,b)
print(random.randint(1,10))
```

```python
#gives a random number in the given range with start stop step system
print(random.randrange(2,20,4))
```

```python
#returns one random item from the given list, tuple or string
alpha = [2,3,4,5,6,7]
print(random.choice(alpha))
```

```python
#returns multiple choices with repetition where k is the keyword parameter i.e:n
alp = [3,1,2,5,6,7]
print(random.choices(alp, k=8))
```

```python
#returns random choices without repetition where we don't use k and the choices can exceed the elements in the data structure
print(random.sample(alp,6))
```
