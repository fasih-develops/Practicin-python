```python
import random
#returns a random number betweeen 0.1 and 1 by default
print(random.random())
```

```python
#returns a random  decimal number betweeen a given interval (a,b)
print(random.uniform(2,10))
```

```python
#returns a random integer between a given interval (a,b)
print(random.randint(1,10))
```

```python
#gives a random number in the given range with start stop step system
print(random.randrange(2,20,4))
```
