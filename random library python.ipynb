```python
import random
#returns a random number betweeen 0.1 and 1 by default
print(random.random())
```

```python
#returns a random  decimal number betweeen a given interval (a,b)
print(random.uniform(2,10))
```

```python
#returns a random integer between a given interval (a,b)
print(random.randint(1,10))
```

```python
#gives a random number in the given range with start stop step system
print(random.randrange(2,20,4))
```

```python
#returns one random item from the given list, tuple or string
alpha = [2,3,4,5,6,7]
print(random.choice(alpha))
```

```python
#returns multiple choices with repetition where k is the keyword parameter i.e:n
alp = [3,1,2,5,6,7]
print(random.choices(alp, k=28))
```

```python
#returns random choices without repetition where we don't use k and the choices can exceed the elements in the data structure
print(random.sample(alp,2))
```

```python
#shuffles the items in a given data structure and doesn't retruns anything
print(random.shuffle(alp))
print(alp)
```

```python
#seed initializes the number generator  with a starting value so the  same seed value always gives the same random results
#fixes the seed value to give the same random results 
random.seed(10)
print(random.random())
print(random.random())
print(random.random())
print(random.random())
print(random.uniform(1,100))
print(random.randrange(1,100))
print(random.randint(1,100))
```

```python
#saves the current internal state of python random generator
random.seed(42)
print(random.random())
state = random.getstate()
print(random.random())
print(random.random())
```

```python
#returns the saved current internal state of the python random generator
#pehle state ko save kia phir us ke bad ek adh ouputs liay phir dubara state ko recall kia or wahi outputs obtain kiay
random.setstate(state)
print(random.random())
print(random.random())
```

```python
#returns a non negative number that exactly contains k bits
print(random.getrandbits(31))
```

```python
#generates n random bytes and returns them as a bytes object
print(random.randbytes(16))
```

```python
#returns a floating point number from the given gaussian distribution with the given mean and standard deviation
#gauss distribution : highest probability at the centre, symmetrical on both sides, has a bell type shaped
print(random.gauss(0,4))
#it has uniform distribution all over, every item has the same value in a given range
print(random.uniform(1,10))
#same as gauss
print(random.normalvariate(0,4))
#in triangular distribution some values are more likely than others but the probability changes linearly (min,max,most likely)
print(random.triangular(1,100,50))
#in exponential distribution small values mostly occur large values are rare
print(random.expovariate(1))
#beta distribution returns values between 0 and 1 always where alpha and beta must always be greater than 0
print(random.betavariate(2,5))
#gamma distribution models positive (greater than zero) quantities especially the waiting times until the multiple random events occur
print(random.gammavariate(2,3))
#log normal distribution is a probability distribution where the logarithm of the  values is normally distributed
print(random.lognormvariate(1,3))
#von mises distribution is used to return a random variable from the vonmises distribution, it used for circular measuring and returns one random angle
print(random.vonmisesvariate(1,3))
#pareto distribution occurs where small number of events account for most of the effect
print(random.paretovariate(2))
#weibull distribution is a probability used to model the time until an  event occurs
print(random.weibullvariate(2,3))
```
