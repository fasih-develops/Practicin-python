```python
import tkinter as tk 
import random

root = tk.Tk()
root.title("color game")
score = 0
colors = ["red", "blue", "green", "yellow", "orange", "purple", "pink", "brown", "black", "white", "gray", "cyan", "magenta", "lime", "teal", "violet"]

def new_round():
    global score
    global color_to_guess, random_color
    color_to_guess = random.choice(colors)
    random_color = random.choice(colors)
    label.config(text=color_to_guess, fg=random_color) #bg=color_to_guess)
    entry.delete(0, tk.END)

def check_answer():
    global score
    user_input = entry.get().lower()
    if user_input == random_color:
        score+= 1
        score_label.config(text="Score: " +str(score))
    else:
        score_label.config(text="Score: " +str(score) + "(Incorrect)")
        new_round()

label = tk.Label(root, font=("Arial", 24))
label.pack(pady=20)

entry = tk.Entry(root, font=("Arial", 18))
entry.pack(pady=20)

check_button = tk.Button(root, text="Check", command=check_answer)
check_button.pack(pady=10)

score_label = tk.Label(root, text="Score: 0")
score_label.pack()

new_round()
root.mainloop()
```
